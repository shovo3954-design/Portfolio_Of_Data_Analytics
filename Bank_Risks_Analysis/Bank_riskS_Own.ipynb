```python
categorical_columns = ['Nationality', 'Occupation', 'Fee Structure',
       'Loyalty Classification','Amount of Credit Cards', 'Properties Owned',
       'Risk Weighting', 'BRId', 'GenderId', 'IAId']

for col in categorical_columns:
    print(f"ValueCounts for {col}:")
    print(df[col].value_counts())

```

```python
# 1. Univariate analysis

for col in categorical_columns:
    plt.figure(figsize=(8, 4))
    top_categories = df[col].value_counts().head(15)  # cap at top 15
    sns.barplot(x=top_categories.values, y=top_categories.index, palette='viridis')
    plt.title(f'Value Counts for {col}')
    plt.xlabel('Count')
    plt.ylabel(col)
    plt.tight_layout()
    plt.show()
```

```python
# 2.Bivariate analysis With Gender

for col in categorical_columns:
    if col == 'GenderId':  # skip plotting gender against itself
        continue
    
    plt.figure(figsize=(9, 5))
    
    # get top 15 categories overall (so both genders' bars stay readable)
    top_cats = df[col].value_counts().head(15).index
    subset = df[df[col].isin(top_cats)]
    
    sns.countplot(data=subset, y=col, hue='GenderId', order=top_cats, palette='viridis')
    plt.title(f'{col} by Gender')
    plt.xlabel('Count')
    plt.ylabel(col)
    plt.legend(title='Gender')
    plt.tight_layout()
    plt.show()
```

```python
# 2.Bivariate analysis With Nationality.

for col in categorical_columns:
    if col == 'Nationality':  # skip plotting gender against itself
        continue
    
    plt.figure(figsize=(9, 5))
    
    # get top 15 categories overall (so both genders' bars stay readable)
    top_cats = df[col].value_counts().head(15).index
    subset = df[df[col].isin(top_cats)]
    
    sns.countplot(data=subset, y=col, hue='Nationality', order=top_cats, palette='viridis')
    plt.title(f'{col} by Nationality')
    plt.xlabel('Count')
    plt.ylabel(col)
    plt.legend(title='Nationality')
    plt.tight_layout()
    plt.show()
```

```python
# 2.Bivariate analysis With Loyalty Classification.

for col in categorical_columns:
    if col == 'Loyalty Classification':  # skip plotting loyalty classification against itself
        continue
    
    plt.figure(figsize=(9, 5))
    
    # get top 15 categories overall (so both genders' bars stay readable)
    top_cats = df[col].value_counts().head(15).index
    subset = df[df[col].isin(top_cats)]
    
    sns.countplot(data=subset, y=col, hue='Loyalty Classification', order=top_cats, palette='viridis')
    plt.title(f'{col} by Loyalty Classification')
    plt.xlabel('Count')
    plt.ylabel(col)
    plt.legend(title='Loyalty Classification')
    plt.tight_layout()
    plt.show()
```

```python
for col in Numberical_columns:
    if col == 'Superannuation Savings':
        # Skip descriptive statistics for the binned column
        continue
    if col == 'Estimated Income':
        # Skip descriptive statistics for the 'Estimated Income' column
        continue
    print(f"Descriptive statistics for {col}:")
    print(df[col].describe())
```

```python
bins = [0, 30000, 70000, float('inf')]
lebels = ['Low', 'Medium', 'High']

df['Foreign Currency Account'] = pd.cut(df['Foreign Currency Account'],
                                bins=bins, labels=lebels, right=False)
```

```python
# 1. Univariate analysis

numerical_cols = ['Estimated Income','Foreign Currency Account']

for col in numerical_cols:
    plt.figure(figsize=(8, 4))
    top_categories = df[col].value_counts().head(15)  # cap at top 15
    sns.barplot(x=top_categories.values, y=top_categories.index, palette='viridis')
    plt.title(f'Value Counts for {col}')
    plt.xlabel('Count')
    plt.ylabel(col)
    plt.tight_layout()
    plt.show()
```

```python
# Histogram for categorical columns

for col in categorical_columns:
    plt.figure(figsize=(8, 4))
    sns.histplot(df[col], bins=15, kde=True, color='blue')
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.show()
```

```python
# Univariate analysis and visualization
plt.figure(figsize=(15,15))
for i, col in enumerate(Numberical_columns):
    plt.subplot(3, 3, i+1)
    sns.histplot(df[col], kde=True)
    plt.title(col)
plt.tight_layout()
plt.show()
```
